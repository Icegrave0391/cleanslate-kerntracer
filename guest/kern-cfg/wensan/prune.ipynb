```python
import time
start_time = time.perf_counter()
print('Done!')
```

```python
import networkx as nx

kernel_cg_path = "./callgraph.graphml"
k_cg = nx.read_graphml(kernel_cg_path)
k_cg.remove_edges_from(nx.selfloop_edges(k_cg))

def prune_cycles_fast(G: nx.DiGraph):
    visited = set()
    in_stack = set()
    back_edges = []
    def dfs(u):
        visited.add(u)
        in_stack.add(u)
        for v in G.successors(u):
            if v not in visited:
                dfs(v)
            elif v in in_stack:
                back_edges.append((u, v))
        in_stack.remove(u)
    for node in G:
        if node not in visited:
            dfs(node)
    G.remove_edges_from(back_edges)

prune_cycles_fast(k_cg)

nx.write_graphml(k_cg, "./pruned_callgraph.graphml")

print('Done!')
```

```python
end_time = time.perf_counter()
elapsed = end_time - start_time
hours   = int(elapsed // 3600)
minutes = int((elapsed % 3600) // 60)
seconds = elapsed % 60

print(f"Elapsed time: {hours}h {minutes}m {seconds:.6f}s")
```
